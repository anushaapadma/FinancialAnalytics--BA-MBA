# Historical Financial Data using Python

Financial statement analysis often requires information from multiple years rather than a single reporting period.

Historical financial data allows analysts to:

- Study trends in company performance.
- Measure growth in revenue and profitability.
- Analyse changes in assets, liabilities and cash flows.
- Forecast future financial performance.
- Apply valuation techniques such as Discounted Cash Flow (DCF).

In this notebook, we will download the historical financial statements of Reliance Industries Limited using the **Yahoo Finance API (yfinance)**.

The same approach can later be applied to any listed company by changing its ticker symbol.

```python
# ============================================================
# Create a connection to Reliance Industries Limited.
#
# Every listed company has a unique ticker symbol.
#
# RELIANCE.NS represents Reliance Industries listed on the
# National Stock Exchange (NSE) of India.
#
# The Ticker object provides access to the company's
# financial statements and market information.
# ============================================================

company = yf.Ticker("RELIANCE.NS")
```

```python
# ============================================================
# Download the company's financial statements.
#
# Yahoo Finance provides three major financial statements:
#
# 1. Income Statement
# 2. Balance Sheet
# 3. Cash Flow Statement
#
# Each statement contains information for multiple
# financial years.
# ============================================================

income = company.financials

balance = company.balance_sheet

cashflow = company.cashflow
```

```python
# ============================================================
# Display the downloaded financial statements.
#
# Notice that each column represents a different
# financial year.
#
# These historical values will later be used for
# financial analysis, forecasting and valuation.
# ============================================================

income
```

```python
# ============================================================
# Explore the structure of the Income Statement.
#
# The Income Statement is stored as a Pandas DataFrame.
#
# • Each ROW represents a financial variable.
# • Each COLUMN represents a financial year.
#
# Before extracting any financial information, it is useful
# to examine the available variables reported by the company.
#
# The index attribute displays the names of all available
# financial variables.
#
# These names will later be used with the loc[] method to
# retrieve specific financial variables.
# ============================================================

income.index
```

```python
# ============================================================
# Extract Total Revenue.
#
# The loc[] method is used to retrieve a specific row from
# a Pandas DataFrame using its row label.
#
# Syntax:
#
# dataframe.loc["Row Name"]
#
# Here,
#
# • income      → Income Statement DataFrame
# • loc[]       → Retrieves a row using its label
# • "Total Revenue" → Name of the required financial variable
#
# The extracted result is stored as a Pandas Series.
#
# A Pandas Series is a one-dimensional data structure that
# consists of:
#
# • An index
# • Corresponding values
#
# In this example,
#
# • The index represents the financial years.
# • The values represent the Revenue reported during
#   each financial year.
#
# Since the data spans multiple financial years,
# Revenue is called a Historical Time Series.
# ============================================================

revenue = income.loc["Total Revenue"]

revenue
```

```python
# ============================================================
# Extract additional variables from the Income Statement.
#
# The same loc[] method can be used to retrieve any
# financial variable available in the Income Statement.
#
# Each extracted variable is stored as a Pandas Series,
# containing historical values across multiple financial
# years.
#
# These variables will be used throughout this course for:
#
# • Financial Statement Analysis
# • Ratio Analysis
# • Financial Forecasting
# • Time Value of Money (TVM)
# • Discounted Cash Flow (DCF) Valuation
# • Business Valuation
# ============================================================

gross_profit = income.loc["Gross Profit"]

operating_income = income.loc["Operating Income"]

ebit = income.loc["EBIT"]

ebitda = income.loc["EBITDA"]

net_income = income.loc["Net Income"]
```

```python
# ============================================================
# Display the extracted financial variables.
#
# Notice that each variable is stored as a Pandas Series.
#
# Every Series contains:
#
# • Financial years as the index.
# • Financial values corresponding to each year.
#
# Together, these Series represent the historical
# financial performance of the company.
# ============================================================

print("Revenue")
display(revenue)

print("Gross Profit")
display(gross_profit)

print("Operating Income")
display(operating_income)

print("EBIT")
display(ebit)

print("EBITDA")
display(ebitda)

print("Net Income")
display(net_income)
```

```python
# ============================================================
# Create a Historical Income Statement Dataset.
#
# Each extracted financial variable is currently stored as a
# separate Pandas Series.
#
# For financial analysis, it is more convenient to combine
# these variables into a single DataFrame.
#
# Pandas automatically aligns the Series using the financial
# year as the common index.
#
# Therefore,
#
# • Each row represents one financial year.
# • Each column represents one financial variable.
#
# This creates a structured historical Income Statement
# dataset.
# ============================================================

income_df = pd.DataFrame({

    "Revenue": revenue,
    "Gross Profit": gross_profit,
    "Operating Income": operating_income,
    "EBIT": ebit,
    "EBITDA": ebitda,
    "Net Income": net_income

})

income_df
```

```python
# ============================================================
# Explore the structure of the Balance Sheet.
#
# Similar to the Income Statement, the Balance Sheet is
# stored as a Pandas DataFrame.
#
# • Each row represents a financial variable.
# • Each column represents a financial year.
#
# The index displays all available Balance Sheet variables.
# ============================================================

balance.index
```

```python
# ============================================================
# Extract key Balance Sheet variables.
#
# These variables describe the company's financial position
# at the end of each financial year.
#
# Like the Income Statement variables, each extracted item
# is stored as a historical Pandas Series.
# ============================================================

total_assets = balance.loc["Total Assets"]

current_assets = balance.loc["Current Assets"]

cash = balance.loc["Cash And Cash Equivalents"]

inventory = balance.loc["Inventory"]

accounts_receivable = balance.loc["Accounts Receivable"]

total_liabilities = balance.loc["Total Liabilities Net Minority Interest"]

current_liabilities = balance.loc["Current Liabilities"]

long_term_debt = balance.loc["Long Term Debt"]

total_debt = balance.loc["Total Debt"]

shareholders_equity = balance.loc["Stockholders Equity"]

retained_earnings = balance.loc["Retained Earnings"]
```

```python
# ============================================================
# Create a Historical Balance Sheet Dataset.
#
# The extracted Balance Sheet variables are combined into a
# single DataFrame.
#
# Each row represents a financial year, while each column
# represents a Balance Sheet variable.
# ============================================================

balance_df = pd.DataFrame({

    "Total Assets": total_assets,
    "Current Assets": current_assets,
    "Cash": cash,
    "Inventory": inventory,
    "Accounts Receivable": accounts_receivable,
    "Total Liabilities": total_liabilities,
    "Current Liabilities": current_liabilities,
    "Long Term Debt": long_term_debt,
    "Total Debt": total_debt,
    "Shareholders Equity": shareholders_equity,
    "Retained Earnings": retained_earnings

})

balance_df
```

```python
# ============================================================
# Explore the structure of the Cash Flow Statement.
#
# The Cash Flow Statement explains how cash moves into and
# out of the business during each financial year.
#
# Like the other financial statements:
#
# • Each row represents a financial variable.
# • Each column represents a financial year.
#
# The index displays all available Cash Flow variables.
# ============================================================

cashflow.index
```

```python
# ============================================================
# Extract key Cash Flow variables.
#
# These variables explain the cash generated and used by
# the company's operating, investing and financing
# activities.
#
# Each extracted variable is stored as a historical
# Pandas Series.
# ============================================================

operating_cash_flow = cashflow.loc["Operating Cash Flow"]

investing_cash_flow = cashflow.loc["Investing Cash Flow"]

financing_cash_flow = cashflow.loc["Financing Cash Flow"]

capital_expenditure = cashflow.loc["Capital Expenditure"]

free_cash_flow = cashflow.loc["Free Cash Flow"]
```

```python
# ============================================================
# Create a Historical Cash Flow Dataset.
#
# The extracted Cash Flow variables are combined into a
# single DataFrame.
#
# Each row represents one financial year, while each
# column represents one Cash Flow variable.
# ============================================================

cashflow_df = pd.DataFrame({

    "Operating Cash Flow": operating_cash_flow,
    "Investing Cash Flow": investing_cash_flow,
    "Financing Cash Flow": financing_cash_flow,
    "Capital Expenditure": capital_expenditure,
    "Free Cash Flow": free_cash_flow

})

cashflow_df
```

```python
# ============================================================
# Create the Historical Financial Dataset.
#
# The Income Statement, Balance Sheet and Cash Flow
# Statement contain different aspects of the company's
# financial performance.
#
# These datasets are combined into a single DataFrame
# using the financial year as the common index.
#
# The resulting dataset provides a comprehensive view of
# the company's historical financial performance.
#
# This dataset will be used throughout the remaining
# chapters for:
#
# • Financial Statement Analysis
# • Time Value of Money (TVM)
# • Financial Forecasting
# • Ratio Analysis
# • Discounted Cash Flow (DCF) Valuation
# • Business Valuation
# ============================================================

historical_df = income_df.join(balance_df)

historical_df = historical_df.join(cashflow_df)

historical_df
```

```python
# ============================================================
# Save the Historical Financial Dataset.
#
# The dataset is exported as a CSV file so that it can be
# reused in future chapters without downloading the
# financial statements again.
# ============================================================

historical_df.to_csv("Historical_Financial_Data.csv")

print("Historical_Financial_Data.csv has been saved successfully.")
```
