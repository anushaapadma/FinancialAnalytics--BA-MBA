# Historical Financial Dataset Creation

In the previous notebook, the financial statements of a
single company were explored to understand their structure.

In this notebook, the same process is automated for
multiple companies.

For every company, historical information from the

• Income Statement
• Balance Sheet
• Cash Flow Statement

will be downloaded and combined into one historical
financial dataset.

The final dataset will be used throughout the remaining
chapters of this book.

```python
companies = [

"ADANIENT.NS",
"ADANIPORTS.NS",
"APOLLOHOSP.NS",
"ASIANPAINT.NS",
"AXISBANK.NS",
"BAJAJ-AUTO.NS",
"BAJFINANCE.NS",
"BAJAJFINSV.NS",
"BEL.NS",
"BHARTIARTL.NS",
"CIPLA.NS",
"COALINDIA.NS",
"DRREDDY.NS",
"EICHERMOT.NS",
"ETERNAL.NS",
"GRASIM.NS",
"HCLTECH.NS",
"HDFCBANK.NS",
"HDFCLIFE.NS",
"HEROMOTOCO.NS",
"HINDALCO.NS",
"HINDUNILVR.NS",
"ICICIBANK.NS",
"INDUSINDBK.NS",
"INFY.NS",
"ITC.NS",
"JIOFIN.NS",
"JSWSTEEL.NS",
"KOTAKBANK.NS",
"LT.NS",
"M&M.NS",
"MARUTI.NS",
"NESTLEIND.NS",
"NTPC.NS",
"ONGC.NS",
"POWERGRID.NS",
"RELIANCE.NS",
"SBILIFE.NS",
"SBIN.NS",
"SHRIRAMFIN.NS",
"SUNPHARMA.NS",
"TATACONSUM.NS",
"TATAMOTORS.NS",
"TATASTEEL.NS",
"TCS.NS",
"TECHM.NS",
"TITAN.NS",
"TRENT.NS",
"ULTRACEMCO.NS",
"WIPRO.NS"

]

print(f"Total Companies : {len(companies)}")
```

```python
# ============================================================
# Create an empty list.
#
# The financial information extracted for every company
# and every financial year will be stored in this list.
#
# Later, the list will be converted into a Pandas DataFrame.
# ============================================================

historical_dataset = []
```

```python
# ============================================================
# Create a helper function.
#
# Financial statements are stored as Pandas DataFrames.
#
# This function retrieves an entire historical series for a
# specified financial variable.
#
# If a variable is unavailable for a company, an empty
# Pandas Series is returned instead of generating an error.
# ============================================================

def get_series(statement, variable):

    try:
        return statement.loc[variable]

    except:
        return pd.Series(dtype="float64")
```

```python
# ============================================================
# Download the financial statements of Reliance Industries.
#
# This step verifies that the extraction process works
# correctly before applying it to all companies.
# ============================================================

company = yf.Ticker("RELIANCE.NS")

info = company.info

income = company.financials

balance = company.balance_sheet

cashflow = company.cashflow
```

```python
# ============================================================
# Extract historical financial variables.
#
# Each variable is stored as a historical Pandas Series
# containing values for multiple financial years.
# ============================================================

revenue = get_series(income, "Total Revenue")
gross_profit = get_series(income, "Gross Profit")
operating_income = get_series(income, "Operating Income")
ebit = get_series(income, "EBIT")
ebitda = get_series(income, "EBITDA")
net_income = get_series(income, "Net Income")

total_assets = get_series(balance, "Total Assets")
current_assets = get_series(balance, "Current Assets")
cash = get_series(balance, "Cash And Cash Equivalents")
inventory = get_series(balance, "Inventory")
accounts_receivable = get_series(balance, "Accounts Receivable")

total_liabilities = get_series(balance, "Total Liabilities Net Minority Interest")
current_liabilities = get_series(balance, "Current Liabilities")
long_term_debt = get_series(balance, "Long Term Debt")
total_debt = get_series(balance, "Total Debt")
shareholders_equity = get_series(balance, "Stockholders Equity")
retained_earnings = get_series(balance, "Retained Earnings")

operating_cash_flow = get_series(cashflow, "Operating Cash Flow")
investing_cash_flow = get_series(cashflow, "Investing Cash Flow")
financing_cash_flow = get_series(cashflow, "Financing Cash Flow")
capital_expenditure = get_series(cashflow, "Capital Expenditure")
free_cash_flow = get_series(cashflow, "Free Cash Flow")
```

```python
# ============================================================
# Create the Historical Financial Dataset.
#
# This loop downloads the financial statements of every
# company, extracts historical values for each financial
# year, and stores one record per company per year.
# ============================================================

for symbol in companies:

    print(f"Processing {symbol}")

    try:

        company = yf.Ticker(symbol)

        info = company.info
        income = company.financials
        balance = company.balance_sheet
        cashflow = company.cashflow

        # Company Information
        company_name = info.get("longName")
        ticker = info.get("symbol")
        sector = info.get("sector")
        industry = info.get("industry")

        # Income Statement
        revenue = get_series(income, "Total Revenue")
        gross_profit = get_series(income, "Gross Profit")
        operating_income = get_series(income, "Operating Income")
        ebit = get_series(income, "EBIT")
        ebitda = get_series(income, "EBITDA")
        net_income = get_series(income, "Net Income")

        # Balance Sheet
        total_assets = get_series(balance, "Total Assets")
        current_assets = get_series(balance, "Current Assets")
        cash = get_series(balance, "Cash And Cash Equivalents")
        total_liabilities = get_series(balance, "Total Liabilities Net Minority Interest")
        total_debt = get_series(balance, "Total Debt")
        shareholders_equity = get_series(balance, "Stockholders Equity")

        # Cash Flow Statement
        operating_cash_flow = get_series(cashflow, "Operating Cash Flow")
        investing_cash_flow = get_series(cashflow, "Investing Cash Flow")
        financing_cash_flow = get_series(cashflow, "Financing Cash Flow")
        capital_expenditure = get_series(cashflow, "Capital Expenditure")
        free_cash_flow = get_series(cashflow, "Free Cash Flow")

        # Determine available financial years
        years = revenue.index

        for year in years:

            financial_data = {

                "Company": company_name,
                "Ticker": ticker,
                "Sector": sector,
                "Industry": industry,
                "Financial Year": year.year,

                "Revenue": revenue.get(year),
                "Gross Profit": gross_profit.get(year),
                "Operating Income": operating_income.get(year),
                "EBIT": ebit.get(year),
                "EBITDA": ebitda.get(year),
                "Net Income": net_income.get(year),

                "Total Assets": total_assets.get(year),
                "Current Assets": current_assets.get(year),
                "Cash": cash.get(year),
                "Total Liabilities": total_liabilities.get(year),
                "Total Debt": total_debt.get(year),
                "Shareholders Equity": shareholders_equity.get(year),

                "Operating Cash Flow": operating_cash_flow.get(year),
                "Investing Cash Flow": investing_cash_flow.get(year),
                "Financing Cash Flow": financing_cash_flow.get(year),
                "Capital Expenditure": capital_expenditure.get(year),
                "Free Cash Flow": free_cash_flow.get(year)

            }

            historical_dataset.append(financial_data)

    except Exception as e:

        print(f"Could not process {symbol}")
        print(e)
```

```python
historical_df = pd.DataFrame(historical_dataset)

historical_df.head()
```
