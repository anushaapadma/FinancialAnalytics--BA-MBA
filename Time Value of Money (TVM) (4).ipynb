# 5.1 Present Value (PV)

The historical cash flows of Tata Consultancy Services (TCS) represent the cash generated by the company over different financial years. While these cash flows provide valuable insights into the company's financial performance, they cannot be directly compared because they occur at different points in time.

According to the **Time Value of Money (TVM)** principle, the value of money changes over time. A cash flow received today is worth more than the same cash flow received in the future because today's money can be invested to earn a return.

Therefore, before comparing or evaluating cash flows occurring at different time periods, financial analysts convert them to a common point in time.

One of the most widely used techniques for doing this is **Present Value (PV)**.

Present Value answers the following question:

> **"How much is a future cash flow worth today?"**

## Present Value of Multiple Cash Flows

Suppose an investment is expected to generate cash flows over several years. Since these cash flows occur at different points in time, they cannot be added directly.

To determine their value today, each future cash flow must first be discounted using an appropriate discount rate.

The Present Value (PV) of the cash flows is obtained by adding the discounted values of all future cash flows.

In this section, we will:

1. Create a Python function to calculate the Present Value of multiple cash flows.
2. Test the function using a simple example.
3. Apply the same function to the historical cash flows of Tata Consultancy Services (TCS).

````python
# ============================================================
# Function: find_pv()
#
# Purpose:
# Calculate the Present Value (PV) of a series of future
# cash flows.
#
# Parameters:
# r         : Discount Rate
# cashflows : List of future cash flows
#
# Assumption:
# The first cash flow occurs at the end of Year 1.
# ============================================================

def find_pv(r, cashflows):

    pv = 0

    for i in range(len(cashflows)):
        pv = pv + cashflows[i] / ((1 + r) ** (i + 1))

    return pv
````

# 5.X Future Value (FV)

The Future Value (FV) of money represents the value of an investment or a series of cash flows at a specified point in the future.

While Present Value discounts future cash flows to determine their value today, Future Value compounds cash flows forward to estimate how much they will be worth at a future date.

Future Value is commonly used to answer questions such as:

- How much will an investment be worth after a certain number of years?
- How much will regular cash flows accumulate to in the future?
- What is the future worth of today's investment?

## Future Value of Multiple Cash Flows

Suppose an investment generates cash flows over several years.

Since each cash flow is received at a different point in time, every cash flow must be compounded to the same future date before they can be added together.

The function developed in this section calculates the Future Value of a series of cash flows.

````python
# ============================================================
# Purpose
# ============================================================
#
# Create a function to calculate the Future Value (FV)
# of a series of cash flows.
#
# ============================================================
# Background
# ============================================================
#
# Unlike Present Value, which discounts future cash flows
# back to today's value, Future Value compounds each cash
# flow forward to a common future date.
#
# The first cash flow is compounded for the maximum
# number of years, whereas the last cash flow is
# compounded for zero years because it already occurs at
# the final time period.
#
# ============================================================
# Function Definition
# ============================================================

def find_fv(r, cashflows):

    # Determine the total number of cash flows
    k = len(cashflows)

    # Initialize Future Value
    fv = 0

    # Compound each cash flow to the end of Year k
    for i in range(k):
        fv = fv + cashflows[i] * ((1 + r) ** (k - i - 1))

    # Return the total Future Value
    return fv
````

PV asks: "How far into the future is this cash flow?" So we discount using (i + 1).
FV asks: "How many years remain until the final year?" So we compound using (k - i - 1).

# 5.3 Present Value of an Annuity (PVA)

An annuity is a series of equal cash flows received or paid at
regular intervals over a specified period.

The Present Value of an Annuity (PVA) calculates the value today
of these equal future cash flows.

In this section, we will:
1. Create a Python function to calculate PVA.
2. Test the function using a simple example.
3. Apply the concept to a practical financial scenario.

## PVA Formula

PVA = (A / r) × (1 - 1 / (1 + r)^k)

Where:

A = Equal periodic cash flow
r = Discount rate per period
k = Number of periods

````python
def find_pva(r, k, A):

    pva = (A / r) * (1 - (1 / (1 + r) ** k))

    return pva
````

````python
r = 0.10
k = 5
A = 100

pva = find_pva(r, k, A)

print("Present Value of Annuity =", round(pva, 2))
````

````python
# ============================================================
# TCS Practical Example – Present Value of an Annuity
# ============================================================

# Assumed annual Operating Cash Flow
# for the next 5 years
A = 50000

# Discount rate
r = 0.10

# Number of years
k = 5

# Calculate the Present Value of the annuity
tcs_pva = find_pva(r, k, A)

# Display the result
print("TCS Assumed Annual Cash Flow =", A)
print("Discount Rate =", r)
print("Number of Years =", k)
print("Present Value of TCS Cash Flows =", round(tcs_pva, 2))
````

# 5.4 Future Value of an Annuity (FVA)

The Future Value of an Annuity (FVA) calculates the accumulated
value of a series of equal cash flows at the end of the final
period.

In this section, we will:

1. Create a Python function to calculate FVA.
2. Test the function using a simple example.
3. Apply the concept to a practical financial scenario.

## Future Value of an Annuity Formula

FVA = A × ((1 + r)^k - 1) / r

Where:

A = Equal periodic cash flow
r = Interest rate per period
k = Number of periods

The formula assumes a normal (ordinary) annuity, where payments
occur at the end of each period.

PVA asks: "What are these equal future payments worth today?"
FVA asks: "What will these equal payments accumulate to by the end?

````python
# ============================================================
# Function: find_fva()
#
# Purpose:
# Calculate the Future Value of a normal annuity.
#
# Parameters:
# r : Interest rate per period
# k : Number of periods
# A : Equal cash flow received or paid in each period
#
# ============================================================

def find_fva(r, k, A):

    # Calculate the Future Value of the annuity
    fva = A * (((1 + r) ** k - 1) / r)

    # Return the calculated Future Value
    return fva
````

## Understanding the `find_fva()` Function

The function requires three inputs:

r → Interest rate per period

k → Number of periods

A → Equal cash flow in each period

For example:

r = 0.10
k = 5
A = 100

means that ₹100 is received or invested at the end of each
year for 5 years, at an annual interest rate of 10%.

The function then applies the Future Value of an Annuity formula
and returns the accumulated value at the end of the fifth year.

````python
# ============================================================
# Example: Future Value of an Annuity
# ============================================================

# Annual interest rate
r = 0.10

# Number of periods
k = 5

# Equal annual cash flow
A = 100

# Calculate the Future Value of the annuity
fva = find_fva(r, k, A)

# Display the result
print("Future Value of Annuity =", round(fva, 2))
````

````python
# ============================================================
# TCS Practical Example – Future Value of an Annuity
# ============================================================

# Assumed annual Operating Cash Flow
# for the next 5 years
A = 50000

# Annual interest rate
r = 0.10

# Number of years
k = 5

# Calculate the Future Value of the annuity
tcs_fva = find_fva(r, k, A)

# Display the inputs and result
print("TCS Assumed Annual Cash Flow =", A)
print("Interest Rate =", r)
print("Number of Years =", k)
print("Future Value of TCS Cash Flows =", round(tcs_fva, 2))
````

# 5.5 Present Value of an Annuity Due

An annuity due is a series of equal cash flows where the
cash flow occurs at the beginning of each period.

In a normal (ordinary) annuity, the cash flow occurs at the
end of each period.

The Present Value of an Annuity Due can be calculated by
multiplying the Present Value of a normal annuity by (1 + r).

PVA Due = PVA × (1 + r)

In this section, we will create a Python function to calculate
the Present Value of an Annuity Due.

````python
# ============================================================
# Function: find_pvad()
#
# Purpose:
# Calculate the Present Value of an Annuity Due.
#
# An annuity due receives the cash flow at the beginning
# of each period.
# ============================================================

def find_pvad(r, k, A):

    # First calculate the Present Value of the normal annuity
    pva = find_pva(r, k, A)

    # Since the cash flows occur one period earlier,
    # multiply the normal PVA by (1 + r)
    pvad = pva * (1 + r)

    # Return the Present Value of the annuity due
    return pvad
````

## Example: Present Value of an Annuity Due

Suppose we receive ₹100 at the beginning of every year
for 5 years.

Assume a discount rate of 10%.

We can calculate the Present Value using the `find_pvad()`
function.

````python
# Annual discount rate
r = 0.10

# Number of periods
k = 5

# Equal cash flow
A = 100

# Calculate Present Value of Annuity Due
pvad = find_pvad(r, k, A)

print("Present Value of Annuity Due =", round(pvad, 2))
````

````python
# ============================================================
# TCS Practical Example – Present Value of an Annuity Due
# ============================================================

# Assumed annual cash flow
A = 50000

# Discount rate
r = 0.10

# Number of years
k = 5

# Calculate PVA Due
tcs_pvad = find_pvad(r, k, A)

print("TCS Assumed Annual Cash Flow =", A)
print("Discount Rate =", r)
print("Number of Years =", k)
print("Present Value of TCS Annuity Due =", round(tcs_pvad, 2))
````
