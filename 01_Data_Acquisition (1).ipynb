# Creating a Structured Financial Dataset

The financial statements contain a large number of variables. However, not all variables are required for financial analysis.

In this section, we will extract selected financial variables from the company information, Income Statement, Balance Sheet and Cash Flow Statement and combine them into a structured dataset.

```python
# Extract company information

info = company.info

company_name = info.get("longName")
ticker = info.get("symbol")
sector = info.get("sector")
industry = info.get("industry")

market_cap = info.get("marketCap")
enterprise_value = info.get("enterpriseValue")

pe_ratio = info.get("trailingPE")
pb_ratio = info.get("priceToBook")

eps = info.get("trailingEps")
dividend_yield = info.get("dividendYield")
```

```python
# Extract Income Statement variables

revenue = income.loc["Total Revenue"].iloc[0]

gross_profit = income.loc["Gross Profit"].iloc[0]

operating_income = income.loc["Operating Income"].iloc[0]

ebit = income.loc["EBIT"].iloc[0]

ebitda = income.loc["EBITDA"].iloc[0]

net_income = income.loc["Net Income"].iloc[0]
```

```python
# Extract Balance Sheet variables

total_assets = balance.loc["Total Assets"].iloc[0]

current_assets = balance.loc["Current Assets"].iloc[0]

cash = balance.loc["Cash And Cash Equivalents"].iloc[0]

inventory = balance.loc["Inventory"].iloc[0]

accounts_receivable = balance.loc["Accounts Receivable"].iloc[0]

total_liabilities = balance.loc["Total Liabilities Net Minority Interest"].iloc[0]

current_liabilities = balance.loc["Current Liabilities"].iloc[0]

long_term_debt = balance.loc["Long Term Debt"].iloc[0]

total_debt = balance.loc["Total Debt"].iloc[0]

shareholders_equity = balance.loc["Stockholders Equity"].iloc[0]

retained_earnings = balance.loc["Retained Earnings"].iloc[0]
```

```python
# Extract Cash Flow variables

operating_cash_flow = cashflow.loc["Operating Cash Flow"].iloc[0]

investing_cash_flow = cashflow.loc["Investing Cash Flow"].iloc[0]

financing_cash_flow = cashflow.loc["Financing Cash Flow"].iloc[0]

capital_expenditure = cashflow.loc["Capital Expenditure"].iloc[0]

free_cash_flow = cashflow.loc["Free Cash Flow"].iloc[0]
```

```python
# Create a structured dataset

financial_data = {

    "Company": company_name,
    "Ticker": ticker,
    "Sector": sector,
    "Industry": industry,

    "Market Cap": market_cap,
    "Enterprise Value": enterprise_value,

    "PE Ratio": pe_ratio,
    "PB Ratio": pb_ratio,
    "EPS": eps,
    "Dividend Yield": dividend_yield,

    "Revenue": revenue,
    "Gross Profit": gross_profit,
    "Operating Income": operating_income,
    "EBIT": ebit,
    "EBITDA": ebitda,
    "Net Income": net_income,

    "Total Assets": total_assets,
    "Current Assets": current_assets,
    "Cash": cash,
    "Inventory": inventory,
    "Accounts Receivable": accounts_receivable,

    "Total Liabilities": total_liabilities,
    "Current Liabilities": current_liabilities,
    "Long Term Debt": long_term_debt,
    "Total Debt": total_debt,
    "Shareholders Equity": shareholders_equity,
    "Retained Earnings": retained_earnings,

    "Operating Cash Flow": operating_cash_flow,
    "Investing Cash Flow": investing_cash_flow,
    "Financing Cash Flow": financing_cash_flow,
    "Capital Expenditure": capital_expenditure,
    "Free Cash Flow": free_cash_flow
}
```

In this step, the dictionary is converted into a pandas DataFrame. A DataFrame is a tabular data structure where each row represents an observation and each column represents a variable.

For the current example, the DataFrame contains the financial information for one company.

```python
# ============================================================
# Convert the financial dictionary into a pandas DataFrame.
#
# A DataFrame is a tabular data structure used for storing
# and analyzing data in Python.
#
# In this example, one row represents one company and each
# column represents a financial variable.
# ============================================================

df = pd.DataFrame([financial_data])

df
```

# Automating Data Collection for Multiple Companies

In the previous sections, financial information was extracted for a single company.

Instead of manually repeating the same process for every company, Python allows us to automate data collection using a loop.

In this section, financial information will be extracted for all Nifty 50 companies and combined into a single structured dataset.

```python
# ============================================================
# Create an empty list.
#
# The financial information extracted for each company
# will be stored in this list before creating the final
# structured dataset.
# ============================================================

financial_dataset = []
```

```python
# ============================================================
# Create a helper function.
#
# Some companies may not report every financial variable.
# This function safely extracts a value if it exists.
# Otherwise, it returns None.
# ============================================================

def get_value(statement, variable):
    try:
        return statement.loc[variable].iloc[0]
    except:
        return None
```

```python
# ============================================================
# List of Nifty 50 company ticker symbols.
#
# These ticker symbols will be used to automatically
# download financial information from Yahoo Finance.
# ============================================================

companies = [
    "ADANIENT.NS",
    "ADANIPORTS.NS",
    "APOLLOHOSP.NS",
    "ASIANPAINT.NS",
    "AXISBANK.NS",
    "BAJAJ-AUTO.NS",
    "BAJFINANCE.NS",
    "BAJAJFINSV.NS",
    "BEL.NS",
    "BHARTIARTL.NS",
    "CIPLA.NS",
    "COALINDIA.NS",
    "DRREDDY.NS",
    "EICHERMOT.NS",
    "ETERNAL.NS",
    "GRASIM.NS",
    "HCLTECH.NS",
    "HDFCBANK.NS",
    "HDFCLIFE.NS",
    "HEROMOTOCO.NS",
    "HINDALCO.NS",
    "HINDUNILVR.NS",
    "ICICIBANK.NS",
    "INDUSINDBK.NS",
    "INFY.NS",
    "ITC.NS",
    "JIOFIN.NS",
    "JSWSTEEL.NS",
    "KOTAKBANK.NS",
    "LT.NS",
    "M&M.NS",
    "MARUTI.NS",
    "NESTLEIND.NS",
    "NTPC.NS",
    "ONGC.NS",
    "POWERGRID.NS",
    "RELIANCE.NS",
    "SBILIFE.NS",
    "SBIN.NS",
    "SHRIRAMFIN.NS",
    "SUNPHARMA.NS",
    "TATACONSUM.NS",
    "TATAMOTORS.NS",
    "TATASTEEL.NS",
    "TCS.NS",
    "TECHM.NS",
    "TITAN.NS",
    "TRENT.NS",
    "ULTRACEMCO.NS",
    "WIPRO.NS"
]

print(f"Total companies: {len(companies)}")
```

```python
# ============================================================
# Extract financial information for all companies.
#
# The loop downloads company information, financial
# statements and selected financial variables.
#
# Each company's information is stored as one row in the
# structured financial dataset.
# ============================================================

for symbol in companies:

    print(f"Processing {symbol}")

    try:

        company = yf.Ticker(symbol)

        info = company.info
        income = company.financials
        balance = company.balance_sheet
        cashflow = company.cashflow

        financial_data = {

            "Company": info.get("longName"),
            "Ticker": info.get("symbol"),
            "Sector": info.get("sector"),
            "Industry": info.get("industry"),

            "Market Cap": info.get("marketCap"),
            "Enterprise Value": info.get("enterpriseValue"),

            "PE Ratio": info.get("trailingPE"),
            "PB Ratio": info.get("priceToBook"),
            "EPS": info.get("trailingEps"),
            "Dividend Yield": info.get("dividendYield"),

            "Revenue": get_value(income, "Total Revenue"),
            "Gross Profit": get_value(income, "Gross Profit"),
            "Operating Income": get_value(income, "Operating Income"),
            "EBIT": get_value(income, "EBIT"),
            "EBITDA": get_value(income, "EBITDA"),
            "Net Income": get_value(income, "Net Income"),

            "Total Assets": get_value(balance, "Total Assets"),
            "Current Assets": get_value(balance, "Current Assets"),
            "Cash": get_value(balance, "Cash And Cash Equivalents"),
            "Inventory": get_value(balance, "Inventory"),
            "Accounts Receivable": get_value(balance, "Accounts Receivable"),

            "Total Liabilities": get_value(balance, "Total Liabilities Net Minority Interest"),
            "Current Liabilities": get_value(balance, "Current Liabilities"),
            "Long Term Debt": get_value(balance, "Long Term Debt"),
            "Total Debt": get_value(balance, "Total Debt"),
            "Shareholders Equity": get_value(balance, "Stockholders Equity"),
            "Retained Earnings": get_value(balance, "Retained Earnings"),

            "Operating Cash Flow": get_value(cashflow, "Operating Cash Flow"),
            "Investing Cash Flow": get_value(cashflow, "Investing Cash Flow"),
            "Financing Cash Flow": get_value(cashflow, "Financing Cash Flow"),
            "Capital Expenditure": get_value(cashflow, "Capital Expenditure"),
            "Free Cash Flow": get_value(cashflow, "Free Cash Flow")

        }

        financial_dataset.append(financial_data)

    except Exception as e:

        print(f"Could not process {symbol}")
        print(e)
```

```python
# ============================================================
# Create a structured financial dataset.
#
# The financial information collected for all companies
# is stored as a list of dictionaries. In this step,
# the list is converted into a pandas DataFrame.
#
# Each row represents one company and each column
# represents one financial variable.
# ============================================================

fundamental_df = pd.DataFrame(financial_dataset)

fundamental_df.head()
```
